# Notebook 02 — Validated Pipeline & Statistical Analysis

The complete comparison pipeline.  Every comparison passes through four stages:
1. **Validation** (`src/validation.py`) — reject biologically invalid pairs before computing anything
2. **Metadata extraction** (`src/metadata.py`) — method, resolution, pLDDT per structure
3. **Aligned metrics** (`src/metrics.py`) — all metrics operate on sequence-aligned residues
4. **Statistical analysis** (`src/analysis.py`) — category-level stats and effect sizes

All outputs are written to `results/all_metrics_validated.json` and `results/analysis.json`.

## Comparison categories

| Category | Meaning | Scientific role |
|---|---|---|
| `exp_vs_exp` | X-ray vs NMR, X-ray vs Cryo-EM, NMR vs Cryo-EM | **Baseline variability** — upper bound on what is "normal" |
| `exp_vs_af2` | Any experimental vs AlphaFold2 | **Primary AI comparison** |
| `exp_vs_of3` | Any experimental vs OpenFold3 | Secondary AI comparison (when AF3 data available) |
| `af2_vs_of3` | AlphaFold2 vs OpenFold3 | AI internal comparison |

## Structural scope

All comparisons are at **chain level** (single polypeptide chain, chain A from `proteins.csv`).
This matches the scope of AlphaFold predictions, which are per-UniProt sequence.

### pLDDT context

pLDDT (predicted Local Distance Difference Test) is AlphaFold's per-residue confidence score (0–100).  
AlphaFold documentation defines bands: **<50** very low, **50–70** low, **70–90** confident, **>90** very high.  
We use it to check whether high metric deviation correlates with low confidence regions.


```python
af2_rows = df[df["method_b"] == "AF2"][
    ["protein_name", "plddt_mean_b", "plddt_matched_b", "rmsd"]
].dropna(subset=["plddt_matched_b"])

fig = px.scatter(
    af2_rows,
    x="plddt_matched_b",
    y="rmsd",
    hover_name="protein_name",
    hover_data={"plddt_matched_b": ":.2f", "rmsd": ":.3f"},
    labels={
        "plddt_matched_b": "pLDDT on matched residues (AF2)",
        "rmsd": "RMSD (Å)",
    },
    title="AF2 confidence vs RMSD",
    color_discrete_sequence=[PALETTE["exp_vs_af2"]],
)

# Add confidence band backgrounds
BANDS = [
    (0, 50, "Very low (<50)", "rgba(255,200,200,0.4)"),
    (50, 70, "Low (50–70)", "rgba(255,232,200,0.4)"),
    (70, 90, "Confident (70–90)", "rgba(232,245,233,0.4)"),
    (90, 100, "Very high (>90)", "rgba(227,242,253,0.4)"),
]

for xmin, xmax, label, color in BANDS:
    fig.add_vrect(
        x0=xmin,
        x1=xmax,
        fillcolor=color,
        layer="below",
        line_width=0,
        annotation_text=label,
        annotation_position="top",
        annotation_font_size=10,
    )

fig.update_traces(
    marker=dict(size=10, line=dict(width=1, color="black")),
)
fig.update_layout(
    xaxis=dict(range=[0, 100], title_font_size=12),
    yaxis=dict(range=[0, af2_rows["rmsd"].max() * 1.25], title_font_size=12),
    plot_bgcolor="white",
    width=700,
    height=500,
)

fig.show()
fig.write_html("results/plddt_vs_rmsd.html")
```

## 4. Category-level metric distributions

Visual comparison of metric distributions by category.  
The `exp_vs_exp` distribution is the **baseline** — the natural experimental variability.


```python
present_categories = [c for c in MAIN_CATEGORIES if c in df["category"].values]
metrics_to_plot = [
    m for m in ["rmsd", "t_alpha", "w_rdist_norm", "b_phipsi"] if m in df.columns
]

plot_df = df[df["category"].isin(present_categories)].copy()
cat_x = {c: i for i, c in enumerate(present_categories)}
rng = np.random.default_rng(42)

fig = make_subplots(
    rows=1,
    cols=len(metrics_to_plot),
    subplot_titles=[METRIC_LABEL.get(m, m) for m in metrics_to_plot],
    horizontal_spacing=0.06,
)

added_to_legend = set()

for col, metric in enumerate(metrics_to_plot, start=1):
    sub = plot_df[["protein_name", "category", metric]].dropna()

    for cat in present_categories:
        cat_data = sub[sub["category"] == cat]
        if cat_data.empty:
            continue

        jitter = rng.uniform(-0.18, 0.18, size=len(cat_data))
        x_vals = [cat_x[cat] + j for j in jitter]

        show_legend = cat not in added_to_legend
        if show_legend:
            added_to_legend.add(cat)

        fig.add_trace(
            go.Scatter(
                x=x_vals,
                y=cat_data[metric].values,
                mode="markers",
                name=CATEGORY_LABEL.get(cat, cat).replace("\n", " "),
                legendgroup=cat,
                showlegend=show_legend,
                marker=dict(
                    color=PALETTE[cat],
                    size=9,
                    line=dict(width=0.7, color="black"),
                ),
                text=cat_data["protein_name"].values,
                hovertemplate=(
                    "<b>%{text}</b><br>"
                    + f"{METRIC_LABEL.get(metric, metric)}: "
                    + "%{y:.4f}<br>"
                    + f"Category: {CATEGORY_LABEL.get(cat, cat).replace(chr(10), ' ')}"
                    + "<extra></extra>"
                ),
            ),
            row=1,
            col=col,
        )

        # category mean line
        mean_val = cat_data[metric].mean()
        fig.add_shape(
            type="line",
            x0=cat_x[cat] - 0.35,
            x1=cat_x[cat] + 0.35,
            y0=mean_val,
            y1=mean_val,
            line=dict(color=PALETTE[cat], width=2.2, dash="dash"),
            row=1,
            col=col,
        )

    fig.update_xaxes(
        tickvals=list(cat_x.values()),
        ticktext=[
            CATEGORY_LABEL.get(c, c).replace("\n", " ") for c in present_categories
        ],
        tickfont=dict(size=9),
        row=1,
        col=col,
    )
    fig.update_yaxes(
        rangemode="tozero",
        title_text=METRIC_LABEL.get(metric, metric) if col == 1 else "",
        row=1,
        col=col,
    )

fig.update_layout(
    title=dict(
        text=(
            "Metric distributions by comparison category<br>"
            "<sup>Dot = one comparison pair · Dashed line = category mean · Hover for protein name</sup>"
        ),
        font=dict(size=12),
    ),
    height=520,
    width=430 * len(metrics_to_plot),
    plot_bgcolor="white",
    legend=dict(
        orientation="h",
        y=-0.12,
        x=0.5,
        xanchor="center",
        font=dict(size=10),
    ),
)

fig.show()
fig.write_html("results/category_distributions.html")
```

## 5. Per-protein comparison: all metrics

One subplot per metric.  Within each subplot, each row is one comparison pair — proteins  
are grouped together and coloured by category.  Horizontal bars avoid the rotated-label  
problem from vertical bars; the y-axis labels are always readable.


```python
metrics_to_plot = [
    m for m in ["rmsd", "t_alpha", "w_rdist_norm", "b_phipsi"] if m in df.columns
]

plot_df = df[df["category"].isin(MAIN_CATEGORIES)].copy()
present_cats = [c for c in MAIN_CATEGORIES if c in plot_df["category"].values]

# Build a stable row order: sort by protein, then by category order
cat_order_map = {c: i for i, c in enumerate(MAIN_CATEGORIES)}
plot_df["_cat_order"] = plot_df["category"].map(cat_order_map).fillna(9)
plot_df = plot_df.sort_values(["protein_name", "_cat_order"])

# y-axis label: "Protein  |  X-ray vs AF2"
plot_df["_label"] = (
    plot_df["protein_name"].str.ljust(14)
    + "  |  "
    + plot_df["comparison"].map(COMPARISON_LABEL).fillna(plot_df["comparison"])
)

fig, axes = plt.subplots(
    1,
    len(metrics_to_plot),
    figsize=(4.5 * len(metrics_to_plot), max(5, 0.55 * len(plot_df) + 1.5)),
    sharey=True,  # same y-axis (same rows) across all metric panels
)
if len(metrics_to_plot) == 1:
    axes = [axes]

y_pos = range(len(plot_df))

for ax, metric in zip(axes, metrics_to_plot):
    sub = plot_df[["_label", "category", metric]].copy()

    # draw bars
    bar_colors = [PALETTE.get(cat, "#aaaaaa") for cat in sub["category"]]
    values = sub[metric].fillna(0).values
    bars = ax.barh(
        list(y_pos),
        values,
        color=bar_colors,
        edgecolor="black",
        linewidth=0.5,
        height=0.65,
    )

    # value annotation at end of each bar
    for bar, val in zip(bars, sub[metric]):
        if pd.notna(val) and val > 0:
            ax.text(
                val + 0.004 * (values.max() or 1),
                bar.get_y() + bar.get_height() / 2,
                f"{val:.3f}",
                va="center",
                ha="left",
                fontsize=7.5,
            )

    ax.set_xlabel(METRIC_LABEL.get(metric, metric), fontsize=10)
    ax.set_title(METRIC_LABEL.get(metric, metric), fontsize=11, fontweight="bold")
    ax.set_xlim(left=0, right=(values.max() or 1) * 1.22)
    ax.xaxis.grid(True, linestyle=":", alpha=0.4, zorder=0)
    ax.set_axisbelow(True)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

# y-axis labels only on leftmost subplot
axes[0].set_yticks(list(y_pos))
axes[0].set_yticklabels(plot_df["_label"].tolist(), fontsize=8.5, family="monospace")
axes[0].invert_yaxis()  # top-to-bottom reading order

# protein separator lines — thin horizontal rule between protein groups
protein_groups = plot_df["protein_name"].values
for i in range(1, len(protein_groups)):
    if protein_groups[i] != protein_groups[i - 1]:
        for ax in axes:
            ax.axhline(i - 0.5, color="#cccccc", linewidth=1.0, zorder=0)

# legend
cat_handles = [
    mpatches.Patch(
        facecolor=PALETTE[c],
        edgecolor="black",
        label=CATEGORY_LABEL.get(c, c).replace("\n", " "),
    )
    for c in present_cats
]
fig.legend(
    handles=cat_handles,
    loc="lower center",
    ncol=len(cat_handles),
    fontsize=9,
    bbox_to_anchor=(0.5, -0.06),
    frameon=True,
)

plt.suptitle(
    "All metrics per comparison — experimental baseline vs AI predictions",
    fontsize=12,
    y=1.02,
)
plt.tight_layout()
plt.savefig("results/per_protein_all_metrics.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 6. Effect size summary table

Cohen's d measures the standardised distance between category means.  
Positive d means the AI category has a *larger* mean metric than the exp baseline.  
Values: |d| < 0.5 small, 0.5–0.8 medium, >0.8 large.


```python
es = results["analysis"]["effect_sizes"]
rows = []
for cat, metrics in es.items():
    for m, d in metrics.items():
        if d is not None:
            rows.append({"category": cat, "metric": m, "Cohen's d": round(d, 3)})

if rows:
    es_df = pd.DataFrame(rows).pivot(
        index="metric", columns="category", values="Cohen's d"
    )
    # Rename index and columns for readability
    es_df.index = [METRIC_LABEL.get(m, m) for m in es_df.index]
    es_df.columns = [CATEGORY_LABEL.get(c, c).replace("\n", " ") for c in es_df.columns]
    display(
        es_df.style.background_gradient(cmap="RdBu_r", axis=None, vmin=-2, vmax=2)
        .format("{:.3f}", na_rep="—")
        .set_caption(
            "Cohen's d vs exp_vs_exp baseline  |  red = larger deviation, blue = smaller"
        )
    )
else:
    print("No effect sizes available yet — add more comparison categories.")
```

## 7. Biological interpretation notes

This section documents method-specific interpretation constraints **before** drawing any conclusion.

| Method pair | Interpretation caution |
|---|---|
| X-ray vs NMR | X-ray gives a single static conformation; NMR captures conformational ensemble in solution.  Large RMSD here is **expected** for flexible proteins and is not an error in either structure. |
| X-ray vs Cryo-EM | Cryo-EM resolution varies across proteins and datasets.  Low RMSD between X-ray and Cryo-EM is expected for rigid proteins, confirming near-identical conformation across methods. |
| Exp vs AF2/OF3 | AlphaFold predicts a single structure (highest confidence conformation).  Comparing it to NMR may over-estimate deviation because NMR model 0 is just one conformer from an ensemble.  Compare AF against X-ray (static) for the fairest baseline. |
| RMSD sensitivity | RMSD is dominated by the most deviant regions (flexible loops, termini).  A protein with 3 well-structured helices and 1 flexible loop may have RMSD > 5 Å even if the helices are nearly identical. |
| b_phipsi | Measures distribution similarity, NOT per-residue accuracy.  Two structures can have similar distributions but different residue-level phi/psi assignments (many-to-many permutations). |
| Large X-ray vs NMR RMSD | Some proteins are intrinsically flexible (e.g. dumbbell-shaped or multi-domain proteins).  Large RMSD between X-ray and NMR can reflect real conformational change between calcium-bound/free or bound/unbound states — known biology, not measurement error. |

### Surface metric exclusion

The Machaon paper includes a solvent-accessible surface area (SASA) metric.  
This pipeline excludes it because reliable SASA computation requires all-atom coordinates and an external tool (NACCESS, FreeSASA).  All other metrics here use Cα-only, creating an inconsistent structural basis if SASA were mixed in.  This is a documented limitation; see `src/pipeline.py` docstring.


## 8. Metric limitations reference

| Metric | Limitation |
|---|---|
| RMSD | Sensitive to flexible termini and loops; a single 10 Å outlier dominates. Not invariant to protein size (larger proteins can tolerate larger RMSD). |
| t_alpha | Measures **alpha-shape surface complexity** (triangle count ratio), not coordinate deviation. Sensitive to the alpha=0.085 parameter: highly disordered or very small structures may produce degenerate meshes (0 triangles → `None`).|
| w_rdist_raw | In Ångströms and size-dependent: larger proteins have larger absolute Wasserstein distances. Use `w_rdist_norm` (log10-compressed) for cross-protein comparison. |
| w_rdist_norm | log10 compression reduces dynamic range but does not remove size dependence entirely for very large proteins (>500 residues). |
| b_phipsi | Gaussian assumption on circular embeddings is an approximation; less reliable for proteins with many glycine (Ramachandran outlier-prone) or proline residues. Compares angle distributions, not per-residue angles. |

**The combination of all four metrics provides complementary views:**
- RMSD: coordinate-space backbone deviation (Å)
- t_alpha: gross 3D surface topology (alpha-shape triangle count ratio)
- w_rdist: global shape of pairwise Cα distance distribution (Earth Mover's Distance)
- b_phipsi: backbone dihedral geometry (secondary structure fingerprint)
